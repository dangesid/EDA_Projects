# Line Plot

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,100)
y = np.sin(x)

plt.figure(figsize=(8,6))
plt.plot(x,y,label="Sin Wave",color='blue')
plt.xlabel("X axis")
plt.ylabel("y axis")
plt.grid(True)
plt.legend()
plt.show()
```

# Scatter Plot

```python
import seaborn as sns
import numpy as np

x = np.random.rand(150) #Print random 50 values
y = x + np.random.normal(0,0.1,150) #

plt.figure(figsize=(7,5))
sns.scatterplot(x=x,y=y,color='red')
plt.title("Scatter Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Scatter Plot")
plt.legend()
plt.grid(True)
plt.show()
```

# Bar Chart

```python
categories = ["A", "B", "C", "D"]
values = [10,20,15,25]

plt.figure(figsize=(7,5))
plt.bar(categories,values,color="green")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("bar Chart")
# plt.legend()
plt.grid(True)
plt.show()
```

# Box Plot

```python
# histogram

data = np.random.normal(0,1,1000)

plt.figure(figsize=(7,5))
plt.hist(data,bins=30,color="green")
plt.title("Histogram Chart")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.grid(True)
plt.show()
```

# Box Plot

```python
data = np.random.normal(0,1,100)

plt.figure(figsize=(5,4))
sns.boxplot(data,color='green')
plt.title("BoX Plot")
plt.xlabel("Data Distribution")
plt.ylabel("Values")
# plt.grid(True)
plt.show()
```

# Heat Map

```python
data = np.random.rand(10,10)

plt.figure(figsize=(8,6))
sns.heatmap(data,annot=True,cmap='coolwarm')
plt.title("Heat Map")
plt.show()
```

# Pair Plot

```python
# Sample data
tips = sns.load_dataset('tips')

# Creating a pair plot
sns.pairplot(tips, hue='sex')
plt.suptitle('Pair Plot Example', y=1.02)
plt.show()
```
